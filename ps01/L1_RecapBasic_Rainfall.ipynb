# **Rainfall problem**
Fuzzy spec:
*   Design a program that consumes a list of numbers (integers) representing daily rainfall amounts as entered by a user. 
*   The list may contain the number -999 indicating the end of the data of interest.
*   Produce the average of the non-negative values in the list up to the first -999 (if it shows up). 
*   There may be negative numbers other than-999 in the list.

*If something is not 100% clear: make assumptions and document them!*  (Note: in 'real'-life, don't make assumptions -- ask the client!)

```python
numbers = [1.0, -4, 2.0, 2.5, 0.5, 2, -999, -7, 4]
#numbers = [-999]
numbers
```

```python
#some imports we will use later on

import statistics
import itertools
```

```python
#Not 100% working (doesn't handle a potential div-by-zero case); also bad variable naming
s = 0
c = 0
for i in numbers:
  if i==-999:
    break
  if i>=0:
    s += i
    c += 1
print(s/c)
```

```python
#Variant with for-loop and break; still bad variable naming; handles divByZero with if/else
s = 0
c = 0
for i in numbers:
  if i==-999:
    break
  if i>=0:
    s += i
    c += 1
if c>0:
  print(s/c)
else:
  print("no valid number")
```

```python
#Variant with user input (and a while loop)

def withUserInput():
  s=0
  c=0
  while True:
    i = float(input())
    if i==-999:
      break
    if i>=0:
      s+=i
      c+=1
  if c>0:
    print(s/c)
  else:
    print("no valid number")

#withUserInput()
```

```python
#Variant with temp. second list and a function from the statistics module

posValuesInValidRange = []
for i in numbers:
  if i==-999:
    break
  if i>=0:
    posValuesInValidRange.append(i)

if len(posValuesInValidRange) > 0:
  print(statistics.mean(posValuesInValidRange))
else:
  print("no valid number")

#what are the advantages and disadvantages of this approach?
```

```python
#finally we decide to wrap this div-by-zero-check in a separate function
def printMean(values):
  print( statistics.mean(values) if len(values) > 0 else "no valid number" )

printMean([1,2,3])
printMean([])  
```

```python
#Variant with a function from itertools instead of the list comprehension   (and our printMean)

printMean(list(filter(lambda i: i>=0,itertools.takewhile(lambda i:i!=-999, numbers))))

#just checking for empty lists
printMean(list(filter(lambda i: i>=0, itertools.takewhile(lambda i:i!=-999, []))))
```

```python
#Variant with list comprehension and itertools (and printMean)
printMean([n for n in itertools.takewhile(lambda x: x != -999,numbers) if n >= 0])
```
