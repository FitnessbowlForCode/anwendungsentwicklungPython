```python
a = [70,80,90]
b = a[:]   # the slice operator makes a (shallow) copy

print(a)
print(b)

b[0] = 100
print(a)
print(b)
```

```python
#lets see why it is a shallow copy, not a deep copy
x = [4,5]
a = [70,80,x]
b = a[:]    #this is the same as b = a.copy()

print(a)
print(b)

b[2][0] = 100
print(a)
print(b)
```

```python
#deep copy
import copy

lst1 = ['a','b',[1, 2]]
lst1shallow = lst1.copy()
lst1deep = copy.deepcopy(lst1)

lst1[2][0] = 99

print(lst1)
print(lst1shallow)
print(lst1deep)
```

# Defining your own classes

```python
#Motivation
#manage a list of persons. Each person has a firstname, a lastname, and an age

#variant 1
firstnames=["first1", "first2", "first3"]
lastnames=["last1", "last2", "last3"]
ages = [1,2,3]






#####
#variant 2 (with dictionaries)
p1 = {"firstname":"first1", "lastname":"last1","age":1}
p2 = {"firstname":"first2", "lastname":"last2","age":2}
p3 = {"firstname":"first3", "lastname":"last3","age":3}
persons= [p1,p2,p3]

#how to work with those variants
def printPers(i):
  #variant 1
  print("Hi, my name is {} {}. I am {} years old.".format(firstnames[i],lastnames[i],ages[i]))
  
  #variant 2
  p = persons[i]
  print("Hi, my name is {} {}. I am {} years old.".format(p["firstname"], p["lastname"], p["age"]))

printPers(2)

## how would we update a person? how would we remove a person from the list?   what would it mean to have many more properties (firstname, lastname, ....)?
```

```python
def printPers(p):
  print("Hi, my name is {} {}. I am {} years old.".format(p.firstname, p.lastname, p.age))

printPers(p1)
```

```python
#The idea of classes is not only to store data (group different attributes); Classes also group data and the functions (methods) on this data
#let's move the printPers "procedure" inside the class and call it "method"
class Person:
  def printPers(p):
    print("Hi, my name is {} {}. I am {} years old.".format(p.firstname, p.lastname, p.age))

p1 = Person()
print(p1)
p1.firstname = "first1"
p1.lastname = "last1"
p1.age = 1
#p1.printPers(p1)
p1.printPers()


#Note that we call p1.printPers without a parameter, but the method-definition takes a parameter. Python actually passes the object on which we call the method as a (first) parameter
```

```python
#by convention this parameter is named self
class Person:
  def printPers(self):
    print("Hi, my name is {} {}. I am {} years old.".format(self.firstname, self.lastname, self.age))


p1 = Person()
print(p1)
p1.firstname = "first1"
p1.lastname = "last1"
p1.age = 1
p1.printPers()
```

```python
#what happens, when we call printPers on some other person object?
p2 = Person()
#p2.printPers()    #this will produce an error (there are no attributes for p2 defined yet)
```

```python
#we want each person to have these 3 attributes. how can we ensure this?

class Person:

  def __init__(self):  # declaration of "instance variables: firstname, lastname, age" <- Konstruktor
    self.firstname = "Max"
    self.lastname = "Mustermann"
    self.age = 0

#Note: start AND end with double-underscore: this is something python-internal (don't name your stuff this way)
#Note: __init__ is called the constructor (or initializer). it is called when you create a new object: p1=Person()     (syntax is similar to function call)
```

```python
p1 = Person()
#p1.firstname = "first1"
#p1.lastname = "last1"
#p1.age = 1
printPers(p1)

#p2=Person()
#printPers(p2)

#print(p1.firstname)


```

```python
#now we don't get an error..... all persons will have these three attributes
#but does it make sense, to initialize a person with "Max Musterman"? No, so we have some more parameters

class Person:

  def __init__(self, firstname, lastn, age=0):      
    self.firstname = firstname
    self.lastname = lastn
    self.age = age

  def printPers(self):
    print("Hi, my name is {} {}. I am {} years old.".format(self.firstname, self.lastname, self.age))

p3 = Person("foo", "bar",6)
p3.printPers()

#Note: the constructor has 4 parameters (the first is again the object itself)
```

```python
#we need to provide the constructor parameter
p1 = Person("first1","last1",1)
p2 = Person("first2","last2",2)
printPers(p1)
printPers(p2)

#alternatively, we could use default values for the parameters: def __init__(self, firstname="Max", lastn="Mustermann", age=0):
```

```python
#One last thing...... so far printing a person results in the following (not very informative output):
p = Person("Elvis", "Presley", 88)
print(p)
```

```python

#with the __str__ method inside a class, we can provide a meaningful output (e.g., for debugging purposes)
class Person:

  def __init__(self, firstname, lastn, age):
    self.firstname = firstname
    self.lastname = lastn
    self.age = age

  def printPers(self):
    print("Hi, my name is {} {}. I am {} years old.".format(self.firstname, self.lastname, self.age))    #note we moved printPers into the class and (by convention) use self instead of p

  def __str__(self):
    return "{} {} ({} years)".format(self.firstname, self.lastname, self.age)


p = Person("Elvis", "Presley", 88)
print(p)

```
